```python
### rotation matrix 2x2 

# [1 2] --> rotate 90 degrees
# 
# rotation matrix for 2x2 matrix
# [
#     [0  -1]
#     [1   0]
# ]

### Output -> [2  -1]

vector = [1, 2]
ROTATION_MATRIX = [
                    [0, -1],
                    [1,  0]
                  ]

print("ROTATION_MATRIX: ")
for row in ROTATION_MATRIX:
    for val in row:
        print(val, end=" ")
    print()
```

```python
def rotate_90_degrees(vector ,matrix):
    res = np.dot(vector, matrix)
    print(res)

rotate_90_degrees(vector=vector, matrix=ROTATION_MATRIX)
```
