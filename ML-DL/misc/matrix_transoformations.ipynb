```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
arr = np.array([
    [1, 2],
    [3, 4],
])

arr
```

```python
arr.shape
```

```python
def print_matrix(matrix):
    for row in matrix:
        for val in row:
            print(val, end=" ")
        print()

print_matrix(matrix=arr)
```

```python
vec = np.array([1,1])

# vector matrix mult
'''
1  1 x  1  2  = (1+3) (2+4) = 4 6
        3  4
'''
vec @ arr
```

```python
print_matrix(matrix=arr)
```

```python
### cols as transformed basis vectors

e1 = np.array([1, 0])
e2 = np.array([0, 1])

print(arr @ e1)
print(arr @ e2)
```

```python
def plot_vector(vectors, colors, labels):
    origin = np.array([0, 0])
    for v, c, l in zip(vectors, colors, labels):
        plt.quiver(*origin, *v, angles='xy', scale_units='xy', scale=1, color=c, label=l)
    
    plt.axhline(0)
    plt.axvline(0)
    plt.grid()
    plt.xlim(-5, 5)
    plt.ylim(-5, 5)
    plt.legend()
    plt.gca().set_aspect('equal')
    plt.show()
```

```python
I = np.eye(2)
print_matrix(I)
```

```python
v = np.array([2, 1])

I @ v
```

```python
plot_vector([v, I @ v], ['b', 'g'], ['original v', 'I v'])
```

```python
### scaling

s = np.array([
             [2, 0],
             [0, 0.5]
            ])

plot_vector([v, s @ v], ['b', 'g'], ['v', 'scaled v'])
```

```python
### reflection across the x axis

reflect = np.array([
    [1, 0],
    [0, -1]
])

plot_vector([v, reflect @ v],
            ['b', 'y'],
            ['v', 'reflected v']
            )
```
