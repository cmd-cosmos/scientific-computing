```python
import numpy as np
import sympy as sp
import matplotlib.pyplot as plt

sp.init_printing(use_latex='mathjax')
plt.style.use("ggplot")

print("Env setup complete.")
```

```python
# Symbolic Gradient

x, y = sp.symbols('x y', real=True)
f = x**2 + 3*y**2 + x*y # base function
# upon differentiation | wrt. x --> (2x + y), wrt. y --> (6y + x) 

grad_f = [sp.diff(f, var) for var in (x, y)]
grad_f
```

```python
# base function is unchanged
grad_f_func = sp.lambdify((x, y), grad_f, 'numpy')
grad_f_func(100, 200)
```
