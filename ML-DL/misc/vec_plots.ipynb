```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
x = np.array([1,2])
y = np.array([4,-4])

x, y
```

```python
def plt_vecs(vecs, colors=['r', 'b'], labels=['x', 'y']):
    origin = np.array([[0,0], [0,0]])
    
    for i, j in enumerate(vecs):
        plt.quiver(*origin[i], j[0], j[1], angles='xy', scale_units='xy', scale=1, color=colors[i], label=labels[i])

    plt.xlim(-5, 10)
    plt.ylim(-5, 10)
    plt.grid(True)
    plt.axhline(0, color='black')
    plt.axvline(0, color='black')
    plt.legend()
    plt.show()

plt_vecs([x, y])
```

```python
# dot product

dot = np.dot(x, y)
print(f"dot prod: {dot}")
```

```python
# vector norms

norm_x = np.linalg.norm(x)
norm_y = np.linalg.norm(y)

norm_x, norm_y
```

```python
#cosine similarity

cos_sim = dot / (norm_x * norm_y)
cos_sim
```

```python
# projection of a onto b

proj_scaler = dot / np.dot(y, y)
proj_x_y = proj_scaler * y

proj_x_y
```
