```python
import numpy as np
```

```python
def f(j):
    x, y = j
    return np.array([x + y, x*y, x**2])

def nummerical_jacobian(f, j, h=1e-5):
    n = len(j)
    m = len(f(j)) 
    jacobian = np.zeros((m, n))
    for i in range(n):
        j_step = j.copy()
        j_step[i] += h
        jacobian[:, i] = (f(j_step) - f(j)) / h
    return jacobian

nummerical_jacobian(f, np.array([3.0, 2.0]))

# jacobian =  1  1
#             y  x
#            2x  0
```
