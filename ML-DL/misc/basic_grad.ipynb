```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
def f(x):
    return x**2

def numerical_grad(f, x, h=1e-5):
    return (f(x+h) - f(x)) / h

numerical_grad(f, 3)
# der of f(x) ==> 2x ==> 2(3) = 6
# derivative based on this is 6.000009999951316 
```

```python
def f2(x):
    return 2*x**3 + 100*x

# 6x^2 + 100 ==> 6(2)^2 + 100 ===> 124 
numerical_grad(f2, 2)

# derivative calc numerically = 124.00012000171044
```
