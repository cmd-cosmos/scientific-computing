```python
import pandas as pd
import numpy as np  
import matplotlib.pyplot as plt 
import seaborn as sns   
```

```python
data = {
    "name" : ["Falcon Heavy", "Falcon 9", "Starship Super Heavy", "Dragon"],
    "flights" : [11, 600,11,30],
    "rating" : [4,5,3,3]
}

df = pd.DataFrame(data=data, index=["alpha", "beta", "gamma", "delta"])
df
```

```python
df.head() # prints first 5 rows
```

```python
df.info() # prints the data types for all columns
```

```python
df.describe() # gives stats for numerical cols
```

```python
# accessing cols

df["flights"]
```

```python
df[["name", "rating"]]
```

```python
# loc ---> selecting row by label

df.loc["gamma"]
```

```python
# iloc ---> selecting row by integer idx

df.iloc[0]
```

```python
# filtering data from the data frame
df[df["rating"] > 3]
```

```python
# adding and modifying cols

df["human rated"] = df["rating"] >= 4
df.head()
```

```python
# grouping and aggregation

df.groupby("human rated")["rating"].mean()
```

```python
# visualization

df["rating"].hist()
plt.show()
```

```python
plt.scatter(df["flights"], df["rating"])
plt.xlabel("flights")
plt.ylabel("rating")
```
