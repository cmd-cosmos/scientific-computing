```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
def f(x):
    return x**3 - x - 2

def df(x):
    return 3*x**2 - 1
```

```python
def foo(x):
    return x**2 - 2

def dfoo(x):
    return 2*x
```

```python
def newtonMethod(f, df, x0, tol=1e-6, maxIters=100):
    """
    solve equations of the format f(x) = 0
    """
    x = x0
    for i in range(maxIters):
        dfx = df(x)
        if dfx == 0:
            raise ZeroDivisionError("derivative is 0, newton's method fails")
        
        xNew = x - f(x)/dfx
        print(f"{i}".ljust(3), "|", xNew)
        if abs(xNew - x) < tol:
            return xNew
        x = xNew
    return x
```

```python
newtonMethod(f, df, 2)
```

```python
newtonMethod(foo, dfoo, 2)
```
