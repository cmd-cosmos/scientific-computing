```python
def f(x):
    return x**3 - x - 2

def df(x):
    return 3*x**2 - 1
```

```python
def newtonMethod(x0, tol=1e-6, maxIters=100):
    """
    solve equations of the format f(x) = 0
    """
    x = x0
    for i in range(maxIters):
        xNew = x - f(x)/df(x)
        if abs(xNew - x) < tol:
            return xNew
        x = xNew
    return x
```
